```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('oral_cancer.csv')
df
```

```python
df.info() # extraindo informações do dataset
```

```python
df.describe() # extraindo informações estatísticas do dataset
```

```python
df.isna().sum() # verificando se há valores ausentes 
```

```python
# medidas descritivas
for qual in ['Country', 'Gender', 'Socioeconomic_Status', 'Diagnosis_Stage', 'Treatment_Type', ]:
    print(f'FREQUÊNCIA ABSLT - {qual}')
    display(df.groupby(qual)[qual].value_counts (normalize=True).mul(100).round(2).astype(str) + '%')   

    print(f'FREQUÊNCIA RELATIVA - {qual}')
    display(df.groupby(qual)[qual].value_counts(normalize=True).mul(100).round(2).astype(str) + '%')

    print(f'MODA: {df[qual].mode().tolist()}\n \n')

```

```python
pd.crosstab(df['Gender'], df['Diagnosis_Stage'], normalize='index').mul(100).round(2).astype(str) + '%' # tabela cruzada para verificar a relação entre gênero e estágio do diagnóstico)
```

```python
pd.crosstab(df['Gender'], df['Treatment_Type'], normalize='index').mul(100).round(2).astype(str) + '%' # tabela cruzada para verificar a relação entre gênero e tipo de tratamento
```

```python
pd.crosstab(df['Socioeconomic_Status'], df['Treatment_Type'], normalize='index').mul(100).round(2).astype(str) + '%' # tabela cruzada para verificar a relação entre status socioeconômico e tipo de tratamento
```

```python
pd.crosstab([df['Gender'], df['Tobacco_Use']], df['Alcohol_Use'], normalize='index').mul(100).round(2).astype(str) + '%' # tabela cruzada para verificar a relação entre gênero e uso de tabaco e álcool
```

```python
#Graficos de distribuição de gênero, status socioeconômico e estágio do diagnóstico!!
for x in ['Gender', 'Socioeconomic_Status']:
    plt.figure(figsize=(10,5))
    sns.countplot(x=x, data=df, hue='Diagnosis_Stage', palette='magma')
    plt.title(f'Distribuição de {x}')
    plt.show()
```

```python
tabaco_genero = df.groupby(['Gender', 'Tobacco_Use']).size().unstack()
tabaco_genero.plot(kind = 'line', ax = plt.gca(), marker = 'o', title = 'Uso de Tabaco por Gênero')
plt.xlabel('Gênero')
plt.ylabel('Quantidade')
plt.show()
    
    
```

```python
alcool_genero = df.groupby(['Gender', 'Alcohol_Use']).size().unstack()
alcool_genero.plot(kind='box', ax = plt.gca(), title = 'Uso de Álcool por Gênero', color = 'red')
plt.xlabel('Gênero')
plt.ylabel('Quantidade')
plt.show()
```

```python
fig , axs = plt.subplots(nrows=3, ncols=2, figsize=(15, 15))

plt.subplot(3, 2, 1)
genero = df['Gender'].value_counts()
genero.plot(kind='bar', color='purple')
plt.title('Distribuição de Gênero')
plt.xlabel('Gênero')
plt.ylabel('Contagem')

plt.subplot(3, 2, 2)
status = df['Socioeconomic_Status'].value_counts()
status.plot(kind = 'density', color='magenta', fontsize=12)
plt.title('Distribuição de Status Socioeconômico')
plt.xlabel('Status Socioeconômico')
plt.ylabel('Contagem')

plt.subplot(3, 2, 3)
estagio = df['Diagnosis_Stage'].value_counts()
estagio.plot(kind='bar', color='green')
plt.title('Distribuição de Estágio do Diagnóstico')

plt.subplot(3, 2, 4)
tratamento = df['Treatment_Type'].value_counts()
tratamento.plot(kind='density', color='green', fontsize=12)
plt.title('Distribuição de Tipo de Tratamento')
plt.xlabel('Tipo de Tratamento')
plt.ylabel('Contagem')

plt.subplot(3, 2, 5)
tabaco = df['Tobacco_Use'].value_counts()
tabaco.plot(kind='bar', color='red')
plt.title('Distribuição de Uso de Tabaco')
plt.xlabel('Uso de Tabaco')
plt.ylabel('Contagem')

plt.subplot(3, 2, 6)
alcool = df['Alcohol_Use'].value_counts()
alcool.plot(kind='density', color='red', fontsize=12)
plt.title('Distribuição de Uso de Álcool')
plt.xlabel('Uso de Álcool')
plt.ylabel('Contagem')

plt.tight_layout()
plt.show()
```

```python
#grafico de calor para verificar a correlação entre as variáveis
plt.figure(figsize=(11, 6))
sns.boxenplot(data=df, hue='Diagnosis_Stage', y='Age', palette='magma')
plt.title('Estágio do Diagnóstico por Idade')
plt.show()
```

```python
#boxplot para verificar a distribuição dos valores de idade por gênero e estágio do diagnóstico
plt.figure(figsize=(11,6))
sns.boxplot(x='Gender', y='Age', hue='Diagnosis_Stage', data=df, palette='magma')
plt.title('Distribuição dos Valores de Idade por Gênero e Estágio do Diagnóstico')
plt.show()
```

```python
#histograma para verificar a distribuição dos valores de idade  

plt.figure(figsize=(10,5))
sns.histplot(df['Age'], color='blue')
plt.title('Distribuição dos Valores de Idade')
plt.xlabel('Idade')
plt.ylabel('Contagem')

plt.show()
```

```python
#mapa de calor

plt.figure(figsize=(11, 6))
sns.heatmap(df.select_dtypes(include=[np.number]).corr(), annot=True, cmap='coolwarm')
plt.title('Mapa de Calor das Correlações')
plt.show()
```
