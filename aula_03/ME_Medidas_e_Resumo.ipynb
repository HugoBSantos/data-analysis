# 1) Quer se estudar o número de erros de impressão de um livro. Para isso escolheu-se uma amostra de 50 páginas, encontrando-se o número de erros por página da tabela anexa.
image.png

## Definindo o DataFrame

```python
tabela = {'Erros': [0, 1, 2, 3, 4], 'Frequência': [25, 20, 3, 1, 1]}
df = pd.DataFrame(tabela)
df.head()
```

## a) Qual o número médio de erros por página?
**Resposta: 0.66**

```python
# calculando a média através do seguinte: soma da multiplicação dos erros pela frequência / total da frequência
media = (df['Erros'] * df['Frequência']).sum() / df['Frequência'].sum()
media
```

## b) E o número mediano?
**Resposta: 0.5**

```python
# obtendo uma lista de todos os valores existentes na tabela
listaValores = (df['Erros'].astype(str) * df['Frequência']).tolist()
listaValores = [i for valor in listaValores for i in valor] #acessando um valor que está dentro de uma valor da lista (para separar '0000' em '0', '0', '0', '0')
print(listaValores)
```

```python
# calculando a mediana através da lista de valores
listaValores.sort() #ordenando a lista em caso dela estar desordenada

# verificando se o tamanho da lista é par ou ímpar e definindo sua mediana
if len(listaValores) % 2 == 0: # se for par, soma os dois valores centrais e defini sua média
    valor1 = int(listaValores[int((len(listaValores)/2)) - 1])
    valor2 = int(listaValores[(int(len(listaValores)/2))])
    mediana = (valor1 + valor2) / 2
else: # se for ímpar, a mediana é o valor centrl
    mediana = int(listaValores[int((len(listaValores)/2))])

# ou utilizando o numpy (aí fica muito fácil)
mediananp = np.median(list(map(int, listaValores)))

print(f'Mediana: {mediana} / Mediana Numpy: {mediananp}')
```

## c) Qual é o desvio padrão?
**Resposta: 0.839**

```python
# calculando a variância
variancia = [(int(i) - media)**2 for i in listaValores]
variancia = sum(variancia) / len(listaValores)
variancia
```

```python
# calculando o desvido padrão
std = np.sqrt(variancia)

# ou utilizando o numpy (aí fica muito fácil)
stdnp = np.std(list(map(int, listaValores)))

print(f'Desvio Padrão: {std} / Desvio Padrão Numpy: {stdnp}')
```

## d) Faça uma representação gráfica para a distribuição do número de erros.
**Resposta:**

image.png

```python
plt.bar(df['Erros'], df['Frequência'])
plt.xlabel('Erros'), plt.ylabel('Frequência'), plt.title('Distribuição do Número de Erros')
for i, freq in enumerate(df['Frequência']):
    plt.text(df['Erros'][i], freq + 0.5, str(freq), ha='center', va='bottom', fontsize=8)

plt.show()
```

```python
# medida de dispersão escolhida: variância

mediacasas = (sum(numcasas) / len(numcasas))
# calculando a variância
variancia = [(i - mediacasas)**2 for i in numcasas]
variancia = sum(variancia) / len(numcasas)
variancia
```
