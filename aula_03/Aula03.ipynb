```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
```

```python
df =pd.read_csv("gasolina.csv")
```

```python
df.head(5)
```

```python
para = df[df['ESTADO'] == 'PARA'].copy()
para
```

```python
para.info()
```

```python
type(para['DATA INICIAL'].iloc[0])
```

```python
para['DATA INICIAL'] = pd.to_datetime(para['DATA INICIAL'])
```

```python
para['DATA FINAL'] = pd.to_datetime(para['DATA FINAL'])
```

```python
para['DATA FINAL'].iloc[0].year
```

```python
para['DATA'] = para['DATA FINAL'].apply(lambda x: f'{x.year}/{x.month}')
```

```python
para['DATA']
```

```python
para.info()
```

```python
para['PREÇO MÉDIO REVENDA']
```

```python
para['PREÇO MÉDIO REVENDA'].mean()
```

```python
para['PREÇO MÉDIO REVENDA'].std()
```

```python
fig, ax = plt.subplots()

ax.plot(para['DATA'] , para['PREÇO MÉDIO REVENDA'])

loc = ticker.MultipleLocator(base=50) # this locator puts ticks at regular intervals
ax.xaxis.set_major_locator(loc)

ax.set_xlabel('Data')
ax.set_ylabel('Preço')
```

```python
fig, ax = plt.subplots()

ax.boxplot(para['PREÇO MÉDIO REVENDA'], vert=False, showmeans = True)

ax.set_xlabel('Preço')
```

```python
para[para["PREÇO MÉDIO REVENDA"] > para["PREÇO MÉDIO REVENDA"].median()]
```

```python
fig, ax = plt.subplots()

ax.boxplot(para[para["PREÇO MÉDIO REVENDA"] > para["PREÇO MÉDIO REVENDA"].median()]["PREÇO MÉDIO REVENDA"], vert=False, showmeans = True)

ax.set_xlabel('Preço')
```
