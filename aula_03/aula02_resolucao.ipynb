```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv("machine1.csv")
```

```python
df.head(3)
```

```python
df["date"] = pd.to_datetime(df["date"])
```

```python
df["month"] = df["date"].apply(lambda x: x.month_name())
```

```python
df["year"] = df["date"].apply(lambda x: x.year)
```

```python
df["coffee_name"].unique()
```

```python
df["cash_type"].unique()
```

```python
df["coffee_name"].value_counts()
```

```python
plt.subplots(figsize = (16,10))

sns.countplot(data=df, x = "coffee_name", 
order=['Americano with Milk', 'Latte', 'Americano', 'Cappuccino', 'Cortado','Hot Chocolate', 'Cocoa', 'Espresso'])

plt.show()
```

```python
pd.crosstab(df["coffee_name"], df["cash_type"])
```

```python
fig, ax = plt.subplots(figsize=(16,10))
sns.countplot(data=df, x="month", hue="coffee_name",
order=['January', 'February','March', 'April', 'May', 'June', 'July',
        'August', 'September','October', 'November', 'December'])

plt.show()
```

```python
ano = df[(df["month"]!="March") | (df["year"] != 2025)]
```

```python
fig, ax = plt.subplots(figsize=(16,10))
sns.countplot(data=ano, x="month", hue="coffee_name",
order=['January', 'February','March', 'April', 'May', 'June', 'July',
        'August', 'September','October', 'November', 'December'])

plt.show()
```
